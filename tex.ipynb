```python
import re
import json

# This is the problematic string you provided
response_str = 'text{\n    {\n        "query": "what are some common applications of carbon nanotubes and their composites?",\n        "strategy": "hybrid_search",\n        "filters": [\n            {\n                "field": "application",\n                "values": [\n                    "common",\n                    "carbon nanotubes",\n                    "composites"\n                ]\n            }\n        ]\n    }\n}'

try:
    # Find the index of the first '{'
    start_outer = response_str.find('{')
    
    # Find the index of the second '{' (which starts the real JSON)
    start_inner = response_str.find('{', start_outer + 1)
    
    # Find the index of the last '}'
    end_outer = response_str.rfind('}')
    
    # Find the index of the '}' that closes the *inner* JSON
    # We search for a '}' *before* the last one
    end_inner = response_str.rfind('}', 0, end_outer)
    
    # Slice the string to get the inner JSON object
    json_str = response_str[start_inner:end_inner+1]
    
    print(f"--- Extracted JSON String ---\n{json_str}\n")
    
    # Parse the extracted JSON string
    data = json.loads(json_str)
    
    # Extract the 'strategy'
    strategy = data.get('strategy')
    
    # Extract 'query' (since 'content' is not in this string)
    # You asked for 'content', but the string provides 'query'.
    content_or_query = data.get('query')
    
    print(f"Successfully Extracted:")
    print(f"Strategy: {strategy}")
    # Note: You asked for 'content', but this string has 'query'
    print(f"Content (from 'query' field): {content_or_query}")

except Exception as e:
    print(f"Failed to parse the string: {e}")
```
